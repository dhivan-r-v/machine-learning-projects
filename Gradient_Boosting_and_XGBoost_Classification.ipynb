## Import Necesaary Libraries

```python
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer

from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score
```

## Import Datasets

```python
breast_cancer_data=load_breast_cancer()
breast_cancer_data_df=pd.DataFrame(breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_data_df["Target"]=breast_cancer_data.target
```

## 3.Data Understandings

```python
breast_cancer_data_df.isnull().sum()
```

```python
breast_cancer_data_df.dtypes
```

## 4.Data Preperation

```python
X=breast_cancer_data_df.drop("Target",axis=1)
```

```python
std_scaler=StandardScaler()
scaled_x=pd.DataFrame(std_scaler.fit_transform(X))
scaled_x
```

```python
y=breast_cancer_data_df.Target
y
```

## 5.Model Building

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
gradient_boosting_classifier=GradientBoostingClassifier(n_estimators=50,max_depth=3)
```

```python
x_train,x_test,y_train,y_test=train_test_split(scaled_x,y,test_size=0.20,random_state=123,shuffle=True,stratify=y)
```

## 6. Model Training

```python
gradient_boosting_classifier.fit(x_train,y_train)
```

```python
from datetime import datetime
```

```python
start_time=datetime.now()
gradient_boosting_classifier.fit(x_train,y_train)
end_time=datetime.now()

print("Time taken to train the algorithm:",end_time-start_time)
```

## 7.Model Testing || 8.Model EValuation

### Training Data

```python
y_pred_train=gradient_boosting_classifier.predict(x_train)
print("Confusion_Matrix: \n",confusion_matrix(y_pred_train,y_train))
print("Accuracy_Score:",accuracy_score(y_pred_train,y_train))
print("Precision_Score:",precision_score(y_pred_train,y_train))
print("Recall_Score:",recall_score(y_pred_train,y_train))
```

### Test Data

```python
y_pred_test=gradient_boosting_classifier.predict(x_test)
print("Confusion_Matrix: \n",confusion_matrix(y_pred_test,y_test))
print("Accuracy_Score:",accuracy_score(y_pred_test,y_test))
print("Precision_Score:",precision_score(y_pred_test,y_test))
print("Recall_Score:",recall_score(y_pred_test,y_test))
```
